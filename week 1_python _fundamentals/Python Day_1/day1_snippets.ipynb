```python
# 1_swap_temp.py — swap using a temporary variable
a=5
b=10
temp=a
a=b
b=temp
print ( "a=",a, "b=",b)

```

```python
# 2_swap_no_temp.py — swap without a temp variable (Python trick)
a = 5
b = 10
a, b = b, a
print("a =", a, "b =", b)
```

```python
# 3_swap_three.py — rotate three variables
a, b, c = 1, 2, 3
a, b, c = c, a, b
print("a =", a, "b =", b, "c =", c)
```

```python
# 4. Celsius to Fahrenheit
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(f"{celsius}°C is {fahrenheit}°F")
```

```python
# 5. Fahrenheit to Celsius
fahrenheit = float(input("Enter temperature in Fahrenheit: "))
celsius = (fahrenheit - 32) * 5/9
print(f"{fahrenheit}°F is {celsius}°C")
```

```python
# 6. Celsius to Kelvin
celsius = float(input("Enter temperature in Celsius: "))
kelvin = celsius + 273.15
print(f"{celsius}°C is {kelvin}K")
```

```python
# 7. Kelvin to Celsius
kelvin = float(input("Enter temperature in Kelvin: "))
celsius = kelvin - 273.15
print(f"{kelvin}K is {celsius}°C")
```

```python
#12. Calculator - All operations in one
num1=float(input("Enter first number: "))
oper=input("Enter operator (+,-,*,/):")
num2=float(input("Enter second number: "))
if oper=='+':
    print("Result= ",num1+num2)
elif oper=='-':
    print("Result= ",num1-num2)
elif oper=='*':
    print("Result= ",num1*num2)
elif oper=='/':
    if num2==0:
     print("infinity")
    else:
     print("Result= ",num1/num2)
else:
    print("Invalid Operator")
```
