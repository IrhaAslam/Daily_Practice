```python
#To-Do List App (saves/loads from a file)
import json

def load_tasks():
    try:
        with open("tasks.json", "r") as f:
            return json.load(f)
    except FileNotFoundError:
        return []

def save_tasks(tasks):
    with open("tasks.json", "w") as f:
        json.dump(tasks, f)

def add_task(tasks, task):
    tasks.append(task)
    save_tasks(tasks)
    print(f"Added: {task}")

def view_tasks(tasks):
    if not tasks:
        print("No tasks yet.")
    else:
        for i, task in enumerate(tasks, 1):
            print(f"{i}. {task}")

def delete_task(tasks, index):
    if 0 <= index < len(tasks):
        removed = tasks.pop(index)
        save_tasks(tasks)
        print(f"Removed: {removed}")
    else:
        print("Invalid task number")

# Test it
tasks = load_tasks()
add_task(tasks, "Finish Day 4 Python practice")
add_task(tasks, "Push code to GitHub")
view_tasks(tasks)
delete_task(tasks, 0)
view_tasks(tasks)
```

```python
#Exception Handling (quick add-on)
def safe_divide(a, b):
    try:
        result = a / b
        print("Result:", result)
    except ZeroDivisionError:
        print("Error: Cannot divide by zero")
    except TypeError:
        print("Error: Please enter valid numbers")

safe_divide(10, 2)
safe_divide(10, 0)
```

```python
#Extract Emails Using Regex
import re

sample_text = """
Contact us at support@example.com or sales@company.org.
You can also reach ali.khan123@gmail.com for personal queries.
"""

emails = re.findall(r'[\w\.-]+@[\w\.-]+\.\w+', sample_text)
print("Found emails:")
for email in emails:
    print(email)
```
